```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

df['Product'] = df['A'] * df['B']

print(df)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(4, 3)),
                  columns=['A', 'B', 'C'])

print("DataFrame:")
print(df)

print("Row-wise sum:")
print(df.sum(axis=1))

print("Column-wise sum:")
print(df.sum(axis=0))
```

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

df.loc[1, 'A'] = np.nan
df.loc[2, 'B'] = np.nan
df.loc[3, 'C'] = np.nan

print("Before filling:")
print(df)

df = df.fillna(df.mean())

print("After filling:")
print(df)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df.loc[1, 'A'] = np.nan
df.loc[3, 'C'] = np.nan

print("Before dropping:")
print(df)

df = df.dropna()

print("After dropping:")
print(df)
```

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(10, 100, size=10)
})

print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print(result)
```

```python
df = pd.DataFrame({
    'Product': np.random.choice(['Pen', 'Book', 'Bag'], size=8),
    'Category': np.random.choice(['Stationery', 'Accessories'], size=8),
    'Sales': np.random.randint(100, 1000, size=8)
})

print(df)

result = df.groupby('Category')['Sales'].sum()

print("Total sales by category:")
print(result)
```

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ali', 'John', 'Sara']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 78]
})

result = pd.merge(df1, df2, on='ID')

print(result)
```

```python
index = pd.MultiIndex.from_tuples([
    ('A', 1), ('A', 2), ('B', 1), ('B', 2)
], names=['Category', 'Number'])

df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)

print("Complete DataFrame:")
print(df)

print("Access category A:")
print(df.loc['A'])

print("Access a specific row:")
print(df.loc[('B', 1)])

print("Slice category A to B:")
print(df.loc['A':'B'])
```

```python
df = pd.DataFrame({
    'Category': ['A', 'A', 'A', 'B', 'B', 'B'],
    'SubCategory': ['X', 'Y', 'X', 'X', 'Y', 'Y'],
    'Value': np.random.randint(10, 100, size=6)
})

df = df.set_index(['Category', 'SubCategory'])

print(df)

print("Sum for each Category and SubCategory:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

result = df.map(lambda x: x * 2)

print("Doubled DataFrame:")
print(result)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(6, 3)),
                  columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'],
                     axis=1)

print(df)
```
